## Integers

- Whole numbers **without** a decimal point.

```python
x = 3          # x is int
type(x)        # <class 'int'>
```

## Float

- Numbers **with** a decimal point.

```python
y = 5.0        # float (even though it looks like an integer)
type(y)        # <class 'float'>
```

## Complex numbers

- Numbers with a real and imaginary part (using `j`).

```python
a = 2 + 4j
b = complex(5, 7)    

print(type(a))    # <class 'complex'>
print(b)          # (5+7j)
```

```python
z = 3 - 1j

print(f"\nReal part of (3 - 1j): {z.real}")    # 3.0
print(f"Imaginary part of (3 - 1j): {z.imag}")    # -1.0
```

## String

- A **sequence of characters** enclosed in quotes.

## Python is Dynamically Typed

- You don't need to declare the type of a variable.
- The content you assign determines the type automatically.

```python
x = 3          # x is int
type(x)        # <class 'int'>

x = 5.7        # x is now float (type changed automatically!)
type(x)        # <class 'float'>

%whos
```

```python
print("apple" < "banana")    # True (alphabetical order)
```

## Logical Operators

```python
age = 25
is_adult = not (age < 18)

print("Is adult?", is_adult)          # True
```

## Membership Operators

```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)          # True
print("orange" not in fruits)      # True
```

## Bitwise Operators

```python
num = 15
print(f"{num} is even?", (num & 1) == 0)    # False
```

## Combining Operations

You can chain multiple operations:

```python
a = 5
d = 2.5
V = (a + d) ** 3 / 4
print(V)    # Result stored in V
```

```python
S1 = "Hello"
S2 = "World"
S = S1 + S2
print(f"Concatenation: {S}")    # "HelloWorld"
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a is b:", a is b)          # False (different objects)
print("a is c:", a is c)          # True (same object)
print("a is not b:", a is not b)  # True

print("a == b:", a == b)          # True (same values)
```

## Boolean (bool)

- Represents True or False.

```python
print(bool(None))     # False
print(bool(0.0))      # False
print(bool(1))        # True
print(bool(-1.5))     # True
print(bool("Hello"))  # True
```

## List

- An **ordered**, **mutable** collection of items.

```python
# Creating Lists
a = list((1, 2, 3, 'apple', 4.5))  # Using list() Constructor
print(a)

b = list("Jupyter")
print(b)
```

```python
# Accessing Items
fruits = ["apple", "banana", "cherry"]
print(fruits[0])       # apple
print(fruits[-1])      # cherry
print(fruits[1:3])     # ['banana', 'cherry']
```

```python
# Modifying Lists
fruits = ["apple", "banana", "cherry"]
fruits[1] = "orange"
print(fruits)          # ['apple', 'orange', 'cherry']
```

1. **append()** - Adds an element at the **end** of the list.

```python
a = [1, 2]
a.append(3)
print(a)
```

```python
fruits = ["apple", "banana"]
fruits.append([1, 2])
print(fruits)    # ['apple', 'banana', [1, 2]]
```

> ⚠️ Note: `append()` adds the entire object as a single element.

2. **insert()** - Adds an element at a **specific position**.

```python
fruits = ["apple", "cherry"]
fruits.insert(1, "banana")
print(fruits)    # ['apple', 'banana', 'cherry']
```

```python
# Insert at beginning
fruits.insert(0, "mango")
print(fruits)    # ['mango', 'apple', 'banana', 'cherry']
```

```python
# Insert at end (same as append)
fruits.insert(len(fruits), "grape")
print(fruits)    # ['mango', 'apple', 'banana', 'cherry', 'grape']
```

3. **extend()** - Adds **multiple elements** to the end of the list.

```python
L1 = [1, 2, 3]
L1.extend([4, 5, 6])
print(L1)    # [1, 2, 3, 4, 5, 6]
```

4. **remove(x)** Removes the **first occurrence** of a value.

```python
fruits = ["apple", "banana", "cherry", "banana"]
fruits.remove("banana")
print(fruits)    # ['apple', 'cherry', 'banana']
```

5. **pop(i)** - Removes and returns the **item at index `i`**. Default is the last item.

```python
fruits = ["apple", "banana", "cherry"]

# Pop last item
last = fruits.pop()
print(last)      # cherry
print(fruits)    # ['apple', 'banana']

# Pop by index
first = fruits.pop(0)
print("\n", first)     # apple
print(fruits)    # ['banana']
```

6. **clear()** - Removes **all items** from the list.

```python
fruits = ["apple", "banana", "cherry"]
fruits.clear()
print(fruits)    # []
```

7. **index(x)** - Returns the **index** of the first occurrence.

```python
fruits = ["apple", "banana", "cherry", "banana"]
print(fruits.index("banana"))    # 1

# With start and end
print(fruits.index("banana", 2))    # 3
```

8. **count(x)** - Counts how many times a value appears.

```python
nums = [1, 2, 2, 3, 3, 3, 4]
print(nums.count(2))    # 2
print(nums.count(3))    # 3
print(nums.count(5))    # 0
```

9. **sort()** - Sorts the list **in place** (ascending by default).

```python
nums = [3, 1, 4, 1, 5, 9, 2]
nums.sort()
print(nums)    # [1, 1, 2, 3, 4, 5, 9]
```

```python
# Descending
nums.sort(reverse=True)
print(nums)    # [9, 5, 4, 3, 2, 1, 1]
```

```python
# Sort strings
fruits = ["banana", "apple", "cherry"]
fruits.sort()
print(fruits)    # ['apple', 'banana', 'cherry']
```

```python
# Sorting with `key`
words = ["banana", "apple", "kiwi"]
words.sort(key=len)
print(words)    # ['kiwi', 'apple', 'banana']
```

10. **reverse()** - Reverses the list **in place**.

```python
nums = [1, 2, 3, 4, 5]
nums.reverse()
print(nums)    # [5, 4, 3, 2, 1]
```

```python
nums = [1, 2, 3]
print(nums[::-1])    # [3, 2, 1] (returns new list)
nums.reverse()       # modifies in place
```

11. **copy()** - Creates a **shallow copy** of the list.

```python
# Copy (separate object)
L1 = [1, 2, 3]
L2 = L1.copy()
L2.append(4)

print(L1)    # [1, 2, 3] ✅ Unchanged!
print(L2)    # [1, 2, 3, 4]
```

```python
# Assignment (same object)
L1 = [1, 2, 3]
L2 = L1
L2.append(4)
print(L1)    # [1, 2, 3, 4] ❌ Changed!
```

## Tuple

- An **ordered, immutable** collection of items.

```python
# Creating Tuples
tuple1 = (1, 2, 3, 4, 5)
tuple2 = ("apple", "banana", "cherry")
tuple3 = (1, "Hello", 3.14)    # Mixed types
tuple4 = ()                    # Empty tuple
tuple5 = (5,)                  # Single-element tuple (comma required!)
```

```python
# Accessing Items
t = (1, 2, 3, 4, 5)
print(t[0])       # 1
print(t[-1])      # 5
print(t[1:3])     # (2, 3)
```

## Dictionary

- Stores **key-value pairs**.

```python
#  Creating Dictionaries
d1 = {1: 'Hello', 2: 'Good', 3: 'Morning'}
print(d1)
```

```python
d2 = dict(a="Hello", b="Good", c="Morning")
print(d2)
```

> ⚠️ **Note**: `dict()` with keyword arguments only allows **string keys**.

```python
keys = [1, 2, 3]
values = ['Geeks', 'For', 'Geeks']
d3 = dict(zip(keys, values))
print(d3)
```

```python
d3 = dict(zip(['Geeks', 'For', 'Geeks'], [1, 2, 3]))
print(d3)    # {'Geeks': 3, 'For': 2}
```

> 💡 **Note**: Duplicate keys are overwritten

```python
# Case Sensitivity
d4 = {'GEEKS': 1, 'For': 2, 'Geeks': 3}
print(d4)
```

> Keys are **case-sensitive**.

```python
# Different Types of Values
d5 = {'name': 'Tanmay', 'age': 18}
print(d5)
```

```python
# Different Types of Keys
d6 = {"name": 'Dhanesh', 2025: "year"}
print(d6)
```

```python
# Accessing Dictionary Items 
# Using `[]`
d = {'name': 'Deva', 1: 'Python', (1,2): [1,2,4]}
print(d['name'])      # Deva
print(d[1])           # Python
print(d[(1,2)])       # [1, 2, 4]
```

> ⚠️ **Warning**: Raises `KeyError` if key doesn't exist.

```python
# Using `.get()`
d = {'name': 'Deva', 1: 'Python', (1,2): [1,2,4]}
print(d.get("name"))    # Deva
print(d.get(1))         # Python
print(d.get((1,2)))     # [1, 2, 4]
print(d.get("age"))     # None (no error!)
print(d.get("age", 0))  # 0 (default value)
```

```python
# Adding a New Key-Value
d = {1: 'Hello', '2': 'Good', 3: 'Morning'}
d[4] = "Sampada"
print(d)
```

```python
# Updating an Existing Key
d = {1: 'Hello', '2': 'Good', 3: 'Morning', 4: 'Astha'}
d[3] = "Evening"
print(d)
```

1. **del** - Removes an item using its **key**.

```python
d = {1: 'Hello', '2': 'Good', 3: 'Morning', 4: 'Shrutika'}
del d[3]
print(d)
```

2. **pop()** - Removes the item with the **given key** and returns **its value**.

```python
d = {1: 'Hello', '2': 'Good', 4: 'Shrutika'}
print(d.pop('2'))    # Good (returns value)
print(d)
```

3. **popitem()** - Removes and returns the **last inserted key-value pair**

```python
d = {1: 'Hello', 4: 'Shrutika'}
K, V = d.popitem()
print(f"Key: {K}, Value: {V}")
print(d)
```

4. **clear()** - Removes all items from the dictionary

```python
d = {1: 'Hello', 2: 'Good'}
d.clear()
print(d)    # {}
```

```python
# Iterating Keys
d = {1: 'Hello', '2': 'Good', 'Morning': 3}
for K in d.keys():
    print(f"Key: {K}")

# Iterating Values
for V in d.values():
    print(f"Value: {V}")

# Iterating Key-Value Pairs
for K, V in d.items():
    print(f"Key, Value: {K}, {V}")
```

```python
# Nested Dictionaries
d = {
    1: 'Welcome',
    2: 'To',
    3: {
        'A': 'Harry',
        'B': 'Potter',
        'C': 'And The',
        'D': "Philosopher's",
        'E': 'Stone'
    }
}
print(d)
```

## Set

- Stores **unordered unique elements**.

```python
# Creating Sets
# Using `set()
set1 = set("Hello Good Morning")
print(set1)
print(type(set1))
```

> Duplicates removed, order random.

```python
# Heterogeneous Elements
set2 = {"Hello", 10, 52.7, True}
print(set2)
```

> Output order varies.

```python
# Accessing Set Items
set1 = set(["Hello", "Good", "Morning"])
for i in set1:
    print(i, end=" ")
```

1. **add()** - Adds an element at the **end** of the list.

```python
set1 = {1, 2, 3, 4}
set1.add(5)
print(set1)    # {1, 2, 3, 4, 5}
```

2. **update()** - Adds **multiple element** at the **end** of the list.

```python
set2 = {1, 2, 3, 4}
set2.update([4, 5, 6])
print(set2)    # {1, 2, 3, 4, 5, 6}
```

> Duplicate `4` is ignored (sets store unique elements only)

3. **remove()** - **Raises error** if element doesn't exist.

```python
set1 = {1, 2, 3, 4, 5}
set1.remove(3)
print(set1)    # {1, 2, 4, 5}

# set1.remove(7)  # ❌ KeyError! Element doesn't exist.
```

4. **discard()** - **No error** if element doesn't exist.

```python
set1 = {1, 2, 3, 4, 5}
set1.discard(3)
print(set1)    # {1, 2, 4, 5}

set1.discard(7)  # ✅ No error
print(set1)      # {1, 2, 4, 5}
```

5. **pop()**

```python
set1 = {1, 2, 3, 4, 5}
val = set1.pop()
print(val)     # Random element
print(set1)    # Remaining set
```

6. **clear()**

```python
set1 = {1, 2, 3, 4, 5}
set1.clear()
print(set1)    # set()
```

```python
# Set Operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)    # {1, 2, 3, 4, 5, 6}
print(a & b)    # {3, 4}
print(a - b)    # {1, 2}
print(a ^ b)    # {1, 2, 5, 6}
```

```python
# Set & Frozenset Operations
a = frozenset([1, 2, 3, 4])
b = frozenset([3, 4, 5, 6])
c = a.copy()

print(c)     # frozenset({1, 2, 3, 4})
print(a.union(b))                   # frozenset({1, 2, 3, 4, 5, 6})
print(a.intersection(b))            # frozenset({3, 4})
print(a.difference(b))              # frozenset({1, 2})
print(a.symmetric_difference(b))    # frozenset({1, 2, 5, 6})
```
